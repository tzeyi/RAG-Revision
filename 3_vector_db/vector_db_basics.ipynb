```python
import chromadb

client = chromadb.Client()
```

Create a colelction

```python
collection = client.create_collection("news_articles")
```

Add text documents to the collection

```python
collection.add(
    ids=["id1", "id2", "id3", "id4"],
    documents=[
        "Apple is leading in a smart phone game with iPhone sales up by 35%",
        "Tesla booked a minor profit of 1 billion $ in Q2",
        "Apples are high in fiber, vitamn C, and various antioxidants",
        "SpaceX got NASA contract worth 10 billion $"
    ]
)
```

Fetch documents & embeddings

```python
data = collection.get(
    include=["documents", "embeddings", "metadatas"]
)

# Print embeddings
for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}")
    print("Text: ", data["documents"][i])
    print("Embedding length: ", len(emb))
    print("First 10 values ", emb[:10])
```

Queries

```python
results = collection.query(
    query_texts=["This is a query related to Elon Musk"],
    n_results=2
)
print(results)
```

```python
results = collection.query(
    query_texts=["Ipad", "Tips for improving health"],
    n_results=2
)
print(results)
```
